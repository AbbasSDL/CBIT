```python
#Abstraction: Define an abstract interface and concrete implementations.
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self) -> float:
        pass

class Circle(Shape):
    def __init__(self, r: float): self.r = r
    def area(self) -> float: return 3.14159 * self.r * self.r

class Rectangle(Shape):
    def __init__(self, w: float, h: float): self.w, self.h = w, h
    def area(self) -> float: return self.w * self.h

shapes = [Circle(2), Rectangle(3, 4)]
for s in shapes:
    print(type(s).__name__, "area=", s.area())
    
#Encapsulation: Hide state and expose safe operations.
class Account:
    def __init__(self, balance: float):
        self.__balance = float(balance)  # private

    def deposit(self, amt: float):
        if amt <= 0: raise ValueError("Must deposit positive amount")
        self.__balance += amt

    def withdraw(self, amt: float):
        if amt <= 0: raise ValueError("Must withdraw positive amount")
        if amt > self.__balance: raise ValueError("Insufficient funds")
        self.__balance -= amt
        return amt

    @property
    def balance(self):
        return self.__balance

acct = Account(100)
acct.deposit(50)
try:
    acct.withdraw(200)
except ValueError as e:
    print("Withdraw failed:", e)
print("Balance:", acct.balance)

#Inheritance: Reuse behavior by subclassing.
class Vehicle:
    def start(self): return "Starting engine"
    def stop(self): return "Stopping"

class Car(Vehicle):
    def start(self): return "Car engine roars"

v = Vehicle()
c = Car()
print(v.start(), "|", c.start())

#Polymorphism: Same interface, different implementations.
class Animal:
    def speak(self): return "..."

class Dog(Animal):
    def speak(self): return "Woof"

class Cat(Animal):
    def speak(self): return "Meow"

def announce(a: Animal):
    print(a.speak())

for pet in (Dog(), Cat()):
    announce(pet)   # both use same `announce` interface
```

```python
# Employees examples demonstrating OOP concepts (abstraction, encapsulation, inheritance, polymorphism)
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def calculate_pay(self):
        pass

class HourlyEmployee(Employee):
    def __init__(self, name, hours, rate):
        super().__init__(name)
        self.__hours = hours
        self.__rate = rate
    def calculate_pay(self):
        return self.__hours * self.__rate

class SalariedEmployee(Employee):
    def __init__(self, name, annual_salary):
        super().__init__(name)
        self.__annual_salary = annual_salary
    def calculate_pay(self):
        # monthly pay
        return self.__annual_salary / 12

class Manager(SalariedEmployee):
    def __init__(self, name, annual_salary, bonus=0):
        super().__init__(name, annual_salary)
        self.__bonus = bonus
    def calculate_pay(self):
        return super().calculate_pay() + self.__bonus

def pay_day(employees):
    for e in employees:
        print(f'{e.name}: {e.calculate_pay():.2f}')

# create instances demonstrating encapsulation (private fields) and inheritance
emps = [
    HourlyEmployee('Ana', 160, 15),
    SalariedEmployee('Ben', 60000),
    Manager('Cara', 90000, bonus=500),
]

pay_day(emps)

# trying to access a private attribute will fail (encapsulation):
# print(emps[1].__annual_salary)  # AttributeError if uncommented

# Polymorphism: pay_day accepts any Employee subclass and calls the same method
```

```python
def first_function():
    x = "first level"
    def second_function():
        x = "second level"
        def third_function():
            nonlocal x  # Refers to the nearest enclosing scope's x
            x = "modified at third level"
            print(f"Value of x inside third_function: {x}")
        third_function()
        print(f"Value of x inside second_function: {x}")
    second_function()
    print(f"Value of x inside first_function: {x}")
    
first_function()
```

```python
employee_names = ["Amina", "Bilal", "Clara"]  # global list
print("Before update:", employee_names)

def manage_employees(new_name, promoted_name):
    global employee_names
    promotion_list = []  # local to manage_employees

    def promote_employee(name):
        nonlocal promotion_list
        if name in employee_names:
            employee_names.remove(name)
            promotion_list.append(name)
            print(f"Promoted: {name}")
        else:
            print(f"Cannot promote {name}: not in employee_names")

    employee_names.append(new_name)
    print(f"Added employee: {new_name}")
    promote_employee(promoted_name)
    print("Promotion list inside function:", promotion_list)

manage_employees("Dina", "Bilal")
print("After update:", employee_names)
```

```python
import socket

def start_server():

    # Create a socket object
    server_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

    # Localhost IP address and port number
    host = '127.0.0.7'
    port = 12345

    # Bind the socket to the IP address and port
    server_socket.bind((host, port))

    # Listen for incoming connections
    server_socket.listen(5)

    print("Server is listening...")
    print(f"Server running on {host}:{port}")

    while True:

        # Accept a connection from the client
        client_socket, addr = server_socket.accept()

        print(f"Got a connection from {addr}")

        # Receive data from the client
        data = client_socket.recv(1024).decode('utf-8')

        print(f"Received from client: {data}")

        # Send a response to the client
        response = "Thank you for connecting"
        client_socket.send(response.encode('utf-8'))

        # Close the client connection
        client_socket.close()

        print("Client connection closed.")


# Start the server
start_server()
```

```python
import socket

def start_client():
    # Create a socket object
    client_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

    # Get the local machine name
    host = '127.0.0.1'  # Localhost
    port = 12346

    # Connect to the server
    client_socket.connect((host, port))

    # Send a message to the server
    message = "Hello, Server!"
    client_socket.send(message.encode('utf-8'))

    # Receive a response from the server
    response = client_socket.recv(1024).decode('utf-8')
    print(f"Received from server: {response}")

    # Close the connection
    client_socket.close()

start_client()
```

```python
#pack layout
import tkinter as tk
root = tk.Tk()
root.geometry("300x200")
label1 = tk.Label(root,text = "labell",bg = "red")
label1.pack(fill=tk.BOTH,expand=True)
label2 = tk.Label(root,text="Label2",bg = "blue")
label2.pack(fill=tk.BOTH,expand=True)
root.mainloop()
```

```python
#grid layout
import tkinter as tk
root = tk.Tk()
root.geometry("300x200")
label1 = tk.Label(root,text = "labell",bg = "red")
label1.grid(row=0,column=0)
label2 = tk.Label(root,text="Label2",bg = "blue")
label2.grid(row=1,column=1)
root.mainloop()
```

```python
#place layout
import tkinter as tk
root = tk.Tk()
root.geometry("300x200")
label1 = tk.Label(root,text = "labell",bg = "red")
label1.place(x= 50, y= 50)
label2 = tk.Label(root,text="Label2",bg = "blue")
label2.place(x=150 , y = 200)
root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.title("Combined layout Example")
frame_top = tk.Frame(root)
frame_bottom = tk.Frame(root)
frame_top.pack(side= tk.TOP , fill= tk.BOTH , expand= True)
frame_bottom.pack(side= tk.BOTTOM, fill= tk.BOTH , expand= True)

label1 = tk.Label(frame_top, text = "Top frame - Lable1")
label2 = tk.Label(frame_top, text = "Top frame - Lable2")
label3 = tk.Label(frame_bottom, text = "Bottom frame  - Lable1")
label4 = tk.Label(frame_bottom, text = "bottom frame - Lable2")

label1.pack(side=tk.LEFT)
label2.pack(side=tk.RIGHT)
label3.grid(row=0, column = 0)
label4.grid(row=0, column=1)

root.mainloop()

```

```python
import tkinter as tk

def on_button_click():
    label.config(text="Button Clicked!")

root = tk.Tk()
root.title("Simple GUI")
root.geometry("300x200")

label = tk.Label(root, text="Hello, Tkinter!")
label.pack(pady=20)

button = tk.Button(root, text="Click Me", command=on_button_click)
button.pack(pady=20)

root.mainloop()
```
